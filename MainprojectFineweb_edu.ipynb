### **4.3 Outlier Treatment**

````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

# IQR Method for text_length
Q1 = df["text_length"].quantile(0.25)
Q3 = df["text_length"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_iqr_cleaned = df[(df["text_length"] >= lower_bound) & (df["text_length"] <= upper_bound)]

# Z-score Method for char_length
z_scores = stats.zscore(df["char_length"])
df_z_cleaned = df[(z_scores < 3) & (z_scores > -3)]

# Boxplot visualization
sns.boxplot(x=df["text_length"])
plt.title("Boxplot of Text Lengths (Outlier Detection)")
plt.show()
````

Observation:

IQR Method: Identified extreme outliers in text_length (very long documents beyond the upper whisker).

Z-score Method: Flagged records with unusually high char_length values (>3 standard deviations).

Outliers represent rare, unusually long texts that could distort model training (e.g., skewing sequence length distribution).

Cleaning with IQR is safer for skewed data like text lengths, while Z-score is useful when distributions are closer to normal.

### **4.6 Project-Specific Preprocessing**

NLP

````python
import re
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Clean text
df["clean_text"] = df["text"].apply(lambda x: re.sub(r"[^a-zA-Z\s]", "", str(x).lower()))

# Remove stopwords
df["clean_text"] = df["clean_text"].apply(
    lambda x: " ".join([word for word in x.split() if word not in ENGLISH_STOP_WORDS])
)

# Tokenization
tokenizer = Tokenizer(num_words=5000)
tokenizer.fit_on_texts(df["clean_text"])
sequences = tokenizer.texts_to_sequences(df["clean_text"])

# Padding
padded = pad_sequences(sequences, maxlen=200, padding="post")
````

**7. Model Training**

**7.1 Model Training**


````python
# ============================================================
# 7.1 MODEL TRAINING
# FineWeb-Edu Dataset
# ============================================================

# Install required libraries
!pip install -q datasets transformers accelerate

# ------------------------------------------------------------
# 1. Import libraries
# ------------------------------------------------------------

import torch

from datasets import load_dataset

from transformers import (
    AutoTokenizer,
    GPT2Config,
    GPT2LMHeadModel,
    TrainingArguments,
    Trainer,
    DataCollatorForLanguageModeling
)


# ------------------------------------------------------------
# 2. Load FineWeb-Edu dataset
# ------------------------------------------------------------

dataset = load_dataset(
    "HuggingFaceFW/fineweb-edu",
    name="sample-10BT",
    split="train"
)

print("Dataset loaded successfully.")
print("Total available records:", len(dataset))


# ------------------------------------------------------------
# 3. Shuffle and select 5,000 records
# ------------------------------------------------------------

dataset = dataset.shuffle(seed=42)

dataset = dataset.select(
    range(5000)
)

print("Records selected:", len(dataset))


# ------------------------------------------------------------
# 4. Split into training and validation data
# ------------------------------------------------------------

dataset_split = dataset.train_test_split(
    test_size=0.10,
    seed=42
)

train_dataset = dataset_split["train"]
validation_dataset = dataset_split["test"]

print("Training records:", len(train_dataset))
print("Validation records:", len(validation_dataset))


# ------------------------------------------------------------
# 5. Load GPT-2 tokenizer
# ------------------------------------------------------------

tokenizer = AutoTokenizer.from_pretrained("gpt2")

# GPT-2 does not have a padding token
tokenizer.pad_token = tokenizer.eos_token


# ------------------------------------------------------------
# 6. Tokenize the dataset
# ------------------------------------------------------------

MAX_LENGTH = 256

def tokenize_function(examples):

    return tokenizer(
        examples["text"],
        truncation=True,
        padding="max_length",
        max_length=MAX_LENGTH
    )


train_tokenized = train_dataset.map(
    tokenize_function,
    batched=True,
    remove_columns=train_dataset.column_names
)

validation_tokenized = validation_dataset.map(
    tokenize_function,
    batched=True,
    remove_columns=validation_dataset.column_names
)

print("Tokenization completed.")


# ------------------------------------------------------------
# 7. Create a small GPT-style model
# ------------------------------------------------------------

config = GPT2Config(
    vocab_size=tokenizer.vocab_size,
    n_positions=MAX_LENGTH,
    n_ctx=MAX_LENGTH,
    n_embd=256,
    n_layer=6,
    n_head=8,
    bos_token_id=tokenizer.bos_token_id,
    eos_token_id=tokenizer.eos_token_id,
    pad_token_id=tokenizer.pad_token_id
)

model = GPT2LMHeadModel(config)

print("Model created successfully.")


# ------------------------------------------------------------
# 8. Data collator for causal language modeling
# ------------------------------------------------------------

data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)


# ------------------------------------------------------------
# 9. Training configuration
# ------------------------------------------------------------

training_args = TrainingArguments(
    output_dir="./fineweb_edu_model",

    # Train for 3 epochs
    num_train_epochs=3,

    # Batch size
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,

    # Helps reduce GPU memory usage
    gradient_accumulation_steps=4,

    # Learning rate
    learning_rate=5e-4,

    # Regularization
    weight_decay=0.01,

    # Evaluate after every epoch
    eval_strategy="epoch",

    # Save after every epoch
    save_strategy="epoch",

    # Display training loss after every epoch
    logging_strategy="epoch",

    # Use mixed precision when GPU is available
    fp16=torch.cuda.is_available(),

    # Do not use external logging services
    report_to="none",

    # Reproducibility
    seed=42
)


# ------------------------------------------------------------
# 10. Create Trainer
# ------------------------------------------------------------

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_tokenized,
    eval_dataset=validation_tokenized,
    data_collator=data_collator
)


# ------------------------------------------------------------
# 11. Train the model
# ------------------------------------------------------------

print("\nStarting model training...")

trainer.train()

print("\nModel training completed!")


# ------------------------------------------------------------
# 12. Save the trained model
# ------------------------------------------------------------

trainer.save_model("./fineweb_edu_model")
tokenizer.save_pretrained("./fineweb_edu_model")

print("\nModel saved successfully!")
````

### **7.2 Training / Validation Accuracy**

````python

# ============================================================
# 7.2 TRAINING / VALIDATION ACCURACY
# ============================================================

import torch
import matplotlib.pyplot as plt
from transformers import GPT2LMHeadModel


# ------------------------------------------------------------
# 1. Select device
# ------------------------------------------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)


# ------------------------------------------------------------
# 2. Function to calculate next-token accuracy
# ------------------------------------------------------------

def calculate_accuracy(model, dataset, batch_size=4):

    model.eval()

    total_correct = 0
    total_tokens = 0

    for start in range(0, len(dataset), batch_size):

        batch = dataset[start:start + batch_size]

        input_ids = torch.tensor(
            batch["input_ids"],
            dtype=torch.long
        ).to(device)

        attention_mask = torch.tensor(
            batch["attention_mask"],
            dtype=torch.long
        ).to(device)

        with torch.no_grad():

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

        # Model predicts the next token
        predictions = torch.argmax(
            outputs.logits[:, :-1, :],
            dim=-1
        )

        # Shift labels by one position
        labels = input_ids[:, 1:]

        # Ignore padding tokens
        mask = attention_mask[:, 1:].bool()

        correct = (
            predictions == labels
        ) & mask

        total_correct += correct.sum().item()
        total_tokens += mask.sum().item()

    return total_correct / total_tokens


# ------------------------------------------------------------
# 3. Calculate accuracy for each saved epoch
# ------------------------------------------------------------

training_accuracy = []
validation_accuracy = []
epochs = [1, 2, 3]


for epoch in epochs:

    checkpoint_path = (
        f"./fineweb_edu_model/checkpoint-{epoch}"
    )

    # Find the actual checkpoint directory
    import os
    import glob

    checkpoints = sorted(
        glob.glob("./fineweb_edu_model/checkpoint-*"),
        key=lambda x: int(x.split("-")[-1])
    )

    if epoch > len(checkpoints):
        print(
            f"Checkpoint for epoch {epoch} not found."
        )
        continue

    checkpoint_path = checkpoints[epoch - 1]

    print(
        f"\nLoading checkpoint for Epoch {epoch}..."
    )

    checkpoint_model = GPT2LMHeadModel.from_pretrained(
        checkpoint_path
    )

    checkpoint_model.to(device)

    # Training accuracy
    train_acc = calculate_accuracy(
        checkpoint_model,
        train_tokenized
    )

    # Validation accuracy
    val_acc = calculate_accuracy(
        checkpoint_model,
        validation_tokenized
    )

    training_accuracy.append(train_acc)
    validation_accuracy.append(val_acc)

    print(
        f"Epoch {epoch} - "
        f"Training Accuracy: {train_acc:.4f}"
    )

    print(
        f"Epoch {epoch} - "
        f"Validation Accuracy: {val_acc:.4f}"
    )

    del checkpoint_model

    if torch.cuda.is_available():
        torch.cuda.empty_cache()


# ------------------------------------------------------------
# 4. Accuracy graph
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    epochs[:len(training_accuracy)],
    training_accuracy,
    marker="o",
    linewidth=2,
    label="Training Accuracy"
)

plt.plot(
    epochs[:len(validation_accuracy)],
    validation_accuracy,
    marker="o",
    linewidth=2,
    label="Validation Accuracy"
)

plt.xlabel("Epoch")

plt.ylabel("Next-Token Accuracy")

plt.title(
    "Training and Validation Accuracy"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 5. Display accuracy values
# ------------------------------------------------------------

print("\nAccuracy Results")
print("-" * 40)

for i in range(len(training_accuracy)):

    print(
        f"Epoch {i + 1}: "
        f"Training = {training_accuracy[i]:.4f}, "
        f"Validation = {validation_accuracy[i]:.4f}"
    )
````

### **7.3 Training / Validation Loss**

````python
# ============================================================
# 7.3 TRAINING / VALIDATION LOSS
# ============================================================

import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 1. Get training history from the Trainer
# ------------------------------------------------------------

history = trainer.state.log_history


# ------------------------------------------------------------
# 2. Extract training and validation loss
# ------------------------------------------------------------

training_epochs = []
training_loss = []

validation_epochs = []
validation_loss = []


for record in history:

    # Training loss
    if "loss" in record and "epoch" in record:

        training_epochs.append(
            record["epoch"]
        )

        training_loss.append(
            record["loss"]
        )

    # Validation loss
    if "eval_loss" in record and "epoch" in record:

        validation_epochs.append(
            record["epoch"]
        )

        validation_loss.append(
            record["eval_loss"]
        )


# ------------------------------------------------------------
# 3. Display loss values
# ------------------------------------------------------------

print("Training Loss:")
for epoch, loss in zip(
    training_epochs,
    training_loss
):
    print(
        f"Epoch {epoch:.0f}: {loss:.4f}"
    )


print("\nValidation Loss:")
for epoch, loss in zip(
    validation_epochs,
    validation_loss
):
    print(
        f"Epoch {epoch:.0f}: {loss:.4f}"
    )


# ------------------------------------------------------------
# 4. Plot Training / Validation Loss
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    training_epochs,
    training_loss,
    marker="o",
    linewidth=2,
    label="Training Loss"
)

plt.plot(
    validation_epochs,
    validation_loss,
    marker="o",
    linewidth=2,
    label="Validation Loss"
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title(
    "Training and Validation Loss"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 5. Calculate final validation perplexity
# ------------------------------------------------------------

import math

if len(validation_loss) > 0:

    final_validation_loss = validation_loss[-1]

    perplexity = math.exp(
        final_validation_loss
    )

    print(
        f"\nFinal Validation Loss: "
        f"{final_validation_loss:.4f}"
    )

    print(
        f"Validation Perplexity: "
        f"{perplexity:.4f}"
    )
````

### **8.4 Optimization Techniques**

* **Dropout:** Dropout can be used to reduce overfitting by randomly deactivating a portion of neurons during training. In this project, dropout is kept at the model's default setting.

* **Early Stopping:** Early stopping can prevent unnecessary training when validation performance stops improving. It can be applied by monitoring validation loss and stopping training when no improvement is observed for several evaluations.

* **Data Augmentation:** Traditional data augmentation is not directly applied because FineWeb-Edu consists of natural-language documents. Artificially modifying the text could change its meaning and reduce the quality of the training data.

* **Hyperparameter Tuning:** Hyperparameter tuning was performed by changing the learning rate between experiments. The baseline used a learning rate of **5 × 10⁻⁴**, while Experiment 3 used **1 × 10⁻⁴**. Other parameters such as the model architecture and dataset size were kept consistent for comparison.

* **Fine-Tuning:** The experiments in this project train a small GPT-style model from scratch rather than fine-tuning an existing pretrained language model. Therefore, conventional fine-tuning is not used as the primary training approach.

* **Regularization:** Weight decay of **0.01** is used during training as a regularization technique. It helps reduce excessive dependence on the training data and can help control overfitting.

**Justification:**

These techniques were considered based on their relevance to language-model training and the limited computational resources available for the project. Hyperparameter tuning and weight decay were incorporated into the experiments, while dropout, early stopping, and fine-tuning were treated as potential techniques rather than claiming they were used when they were not. Data augmentation was not applied because the original educational text was retained without artificial modification. This approach keeps the experiments controlled and makes it easier to compare the effect of individual changes on validation performance.

### **9.5 Overfitting / Underfitting**

**Training Performance:**

The training loss and next-token accuracy were monitored across the training epochs. A decrease in training loss indicates that the model is learning patterns from the FineWeb-Edu training data.

**Validation Performance:**

Validation loss and validation accuracy were monitored after each epoch using the validation dataset. These metrics provide an indication of how well the model performs on data that was not used during training.

**Testing Performance:**

A separate test dataset was not used in the current experiment. Therefore, testing performance cannot be directly reported. The validation dataset was used to evaluate the model during training.

**Generalization:**

Generalization was assessed by comparing training and validation performance. If training and validation loss decrease together and their difference remains relatively small, the model is learning patterns that generalize to unseen data. If training loss continues to decrease while validation loss increases, this indicates possible overfitting. If both training and validation performance remain poor, the model may be underfitting.

**Observation:**

The training and validation curves should be examined together to determine whether the model is overfitting or underfitting. Since the experiment uses only 5,000 FineWeb-Edu records and a relatively small model, the limited dataset size should also be considered when interpreting the model's generalization performance.
